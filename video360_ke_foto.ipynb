```python
import cv2, numpy as np, os

VIDEO = "./input/videos/monas-1.mp4"
OUT = "./input//photos/monas"
JUMLAH = 15
```

```python
cap = cv2.VideoCapture(VIDEO)
total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)
jarak = total // JUMLAH

print("jumlah frame :", total)
print("fps          :", fps)
print("durasi       :", round(total / fps, 1), "detik")
print("ambil 1 foto tiap", jarak, "frame")
```

Scoring ketajaman foto / buang yang blur, jika banyak pixel blur. variance blur kecil

```python
def skor_tajam(frame):
    # kecil = cv2.resize(frame, None, fx=0.25, fy=0.25)
    abu = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(abu, cv2.CV_64F).var()
```

```python
os.makedirs(OUT, exist_ok=True)
terbaik, skor_terbaik = None, 0

for i in range(total):
    ok, frame = cap.read()
    if not ok:
        break

    skor = skor_tajam(frame)
    if skor > skor_terbaik:
        terbaik, skor_terbaik = frame, skor

    if (i + 1) % jarak == 0:
        n = (i + 1) // jarak
        cv2.imwrite(f"{OUT}/{n:03d}.png", terbaik)
        print(f"foto {n}/{JUMLAH} disimpan")
        terbaik, skor_terbaik = None, 0

cap.release()
print("done!")
```
